## 1. Визначення сили як фізичної якості людини

### 1.1 Поняття силових якостей

**Силові якості** — це здатність людини долати зовнішній опір або протидіяти йому за рахунок м'язових зусиль. Від рівня їх розвитку залежать досягнення практично у всіх видах спорту.

Методи контролю за силовими якостями мають давню історію. Перші механічні пристрої для вимірювання сили були створені ще у **XVIII столітті**.

### 1.2 Групи показників при контролі силових якостей

При контролі за силовими якостями враховують **три групи показників**:

| № | Група | Показники |
|---|-------|-----------|
| 1 | **Основні** | а) миттєві значення сили у будь-який момент руху, зокрема **максимальна сила**; б) **середня сила** |
| 2 | **Інтегральні** | **Імпульс сили** (площа під кривою F(t)) |
| 3 | **Диференціальні** | **Градієнт сили** (швидкість зміни миттєвих значень сили) |

### 1.3 Фізичні основи

Згідно із **законами механіки**, кінцевий ефект дії сили визначається **імпульсом сили**:

$$I = \int_{t_1}^{t_2} F(t)\, dt$$

Якщо сила постійна:

$$I = F \cdot t$$

**Середня сила** — умовний показник, рівний відношенню імпульсу сили до часу її дії:

$$\bar{F} = \frac{I}{t} = \frac{\int F(t)\, dt}{t}$$

**Градієнт сили** (диференціальний показник) — швидкість наростання сили:

$$G_F = \frac{dF}{dt} \quad \text{або} \quad G_F = \frac{F_{\max}}{t_{F_{\max}}}$$

## 2. Види сили

### 2.1 Класифікація видів сили

| Вид сили | Характеристика | Застосування |
|----------|---------------|-------------|
| **Максимальна сила** | Найвищий рівень напруги, що може розвинути м'яз. Характеризує абсолютну силу | Важка атлетика, боротьба |
| **Вибухова сила** | Здатність досягти максимуму сили за мінімальний час | Стрибки, метання, спринт |
| **Силова витривалість** | Здатність тривалий час підтримувати значні силові зусилля | Веслування, плавання |
| **Відносна сила** | Максимальна сила відносно маси тіла спортсмена | Єдиноборства (вагові категорії) |
| **Динамічна сила** | Сила, що проявляється під час руху | Більшість видів спорту |
| **Статична сила** | Сила, що проявляється без руху (ізометрично) | Важка атлетика, боротьба |

### 2.2 Зони силових якостей при контролі без апаратури

При метанні предметів різної маси виділяють три зони:

- **Зона I** (малі обтяження) → характеризує **швидкісні** якості  
- **Зона II** (середні обтяження) → характеризує **швидкісно-силові** якості  
- **Зона III** (значні обтяження) → характеризує **силові** якості

### 2.3 Показники надійності силових тестів

| Тип тесту | Коефіцієнт надійності $r_{tt}$ |
|-----------|----------------------------|
| Механічні динамометри | 0,60 – 0,80 |
| Градієнти сили (будь-який спосіб) | 0,70 – 0,80 |
| Тензометричні пристрої (максимальна сила) | 0,85 – 0,95 |

## 3. Контроль за рівнем розвитку сили

### 3.1 Способи реєстрації силових якостей

**Спосіб 1 — без вимірювальної апаратури:**
- Оцінка за найбільшою вагою, яку здатний підняти або утримати спортсмен  
- **Прямий спосіб:** максимальна сила = найбільша вага у жимі штанги лежачи тощо  
- **Непрямий спосіб:** стрибки в довжину/висоту, метання набивних м'ячів, підтягування

**Спосіб 2 — з вимірювальними пристроями:**
- **Динамометри** (механічні пружинного типу: ДПУ-1000, ДПУ-2000, ДПУ-5000 Н, похибка ≤ 2 %)  
- **Динамографи** (безперервний запис F(t))  
- **Тензометричні пристрої** (найточніші; дозволяють вимірювати будь-які силові показники)

### 3.2 Типи силовимірювальних установок

| Тип | Принцип | Особливість |
|-----|---------|-------------|
| **Деформаційний** | Вимірює деформацію тіла, до якого прикладена сила | Статичні або квазістатичні зусилля |
| **Інерційний динамограф** | Вимірює прискорення рухомого тіла | Вимірювання сили в русі |

### 3.3 Метрологічні вимоги до вимірювання силових якостей

1. Визначати та стандартизувати **положення тіла (суглоба)** при повторних спробах  
2. Враховувати **довжину сегментів тіла** при вимірюванні моментів сили  
3. Враховувати **напрям вектора сили**  
4. Дотримуватись загальних метрологічних вимог до контролю за фізичною підготовленістю

## Завдання 1. Первинні дані та статистичний аналіз

### Умова:
Проведено **динамометрію кисті** (вимірювання максимальної сили стискання кисті) у групі спортсменів (n = 20). Вимірювання проведено двічі (спроба 1 і спроба 2) для оцінки надійності тесту.

Одиниця вимірювання: **кілограми-сила (кгс)** або **Ньютони (Н)**, 1 кгс = 9,81 Н.

```python
# ============================================================
# ВИХІДНІ ДАНІ
# ============================================================

# Прізвища спортсменів
names = [
    'Андрієнко', 'Бойко', 'Василенко', 'Гнатенко', 'Демченко',
    'Єременко', 'Захаренко', 'Іваненко', 'Коваленко', 'Литвиненко',
    'Марченко', 'Назаренко', 'Олексієнко', 'Петренко', 'Романенко',
    'Савченко', 'Тимченко', 'Удовенко', 'Філіпенко', 'Харченко'
]

# Маса тіла (кг)
body_weight = np.array([
    75, 82, 68, 90, 78,
    85, 72, 88, 76, 81,
    70, 93, 65, 87, 79,
    84, 73, 91, 77, 83
])

# Динамометрія кисті – спроба 1 (кгс)
grip_1 = np.array([
    48, 55, 42, 61, 51,
    58, 46, 63, 49, 54,
    44, 67, 40, 59, 52,
    57, 47, 65, 50, 56
])

# Динамометрія кисті – спроба 2 (кгс)
grip_2 = np.array([
    47, 56, 43, 60, 52,
    59, 45, 62, 50, 55,
    45, 66, 41, 60, 53,
    58, 48, 64, 51, 55
])

# Краща спроба (max із двох)
grip_best = np.maximum(grip_1, grip_2)

# Відносна сила (кгс / кг тіла)
relative_strength = grip_best / body_weight

# Формування таблиці даних
df = pd.DataFrame({
    'Спортсмен':       names,
    'Маса тіла (кг)':  body_weight,
    'Спроба 1 (кгс)':  grip_1,
    'Спроба 2 (кгс)':  grip_2,
    'Краща спроба (кгс)': grip_best,
    'Відносна сила':   relative_strength.round(3)
})

print("=" * 75)
print("ТАБЛИЦЯ 1. Результати динамометрії кисті (n = 20 спортсменів)")
print("=" * 75)
print(df.to_string(index=False))
print("=" * 75)
```

### 1.1 Розрахунок основних статистичних показників

Для характеристики вибірки розраховують:

| Показник | Формула |
|----------|--------|
| Середнє арифметичне | $\bar{x} = \dfrac{\sum x_i}{n}$ |
| Середнє квадратичне відхилення | $\sigma = \sqrt{\dfrac{\sum(x_i - \bar{x})^2}{n-1}}$ |
| Коефіцієнт варіації | $V = \dfrac{\sigma}{\bar{x}} \cdot 100\%$ |
| Похибка середнього | $m = \dfrac{\sigma}{\sqrt{n}}$ |
| Мінімум, максимум, розмах | $W = x_{\max} - x_{\min}$ |

```python
def calc_stats(data, label):
    """Розраховує та виводить описову статистику."""
    n    = len(data)
    mean = np.mean(data)
    std  = np.std(data, ddof=1)
    cv   = (std / mean) * 100
    sem  = std / np.sqrt(n)
    mn   = np.min(data)
    mx   = np.max(data)
    w    = mx - mn
    med  = np.median(data)
    return {
        'Показник': label,
        'n':        n,
        'M (середнє)':  round(mean, 2),
        'σ (відхилення)': round(std, 2),
        'V% (варіація)':  round(cv, 1),
        'm (похибка)':    round(sem, 2),
        'Min':            round(mn, 2),
        'Max':            round(mx, 2),
        'W (розмах)':     round(w, 2),
        'Медіана':        round(med, 2)
    }

stats_grip1  = calc_stats(grip_1, 'Динамометрія кисті – спроба 1 (кгс)')
stats_grip2  = calc_stats(grip_2, 'Динамометрія кисті – спроба 2 (кгс)')
stats_best   = calc_stats(grip_best, 'Динамометрія кисті – краща спроба (кгс)')
stats_rel    = calc_stats(relative_strength, 'Відносна сила (кгс/кг)')

stats_df = pd.DataFrame([stats_grip1, stats_grip2, stats_best, stats_rel])

print("\n" + "=" * 90)
print("ТАБЛИЦЯ 2. Описова статистика показників силової підготовленості")
print("=" * 90)
print(stats_df.set_index('Показник').T.to_string())
print("=" * 90)

M   = np.mean(grip_best)
sig = np.std(grip_best, ddof=1)
n   = len(grip_best)

print(f"\nОсновні параметри кращої спроби:")
print(f"  M  = {M:.2f} кгс")
print(f"  σ  = {sig:.2f} кгс")
print(f"  V% = {(sig/M*100):.1f}%  → ", end="")
if sig/M*100 < 10:
    print("однорідна група (V < 10%)")
elif sig/M*100 < 20:
    print("помірно однорідна група (10% ≤ V < 20%)")
else:
    print("неоднорідна група (V ≥ 20%)")
```

## Завдання 2. Побудова нормувальної шкали оцінки

У спортивній метрології для оцінки рівня розвитку фізичних якостей використовують **п'ятирівневу нормувальну шкалу** на основі сигмальних відхилень:

| Рівень | Межі | Оцінка |
|--------|------|--------|
| Низький | $x < M - 1{,}5\sigma$ | 1 (незадовільно) |
| Нижче середнього | $M - 1{,}5\sigma \leq x < M - 0{,}5\sigma$ | 2 (задовільно) |
| Середній | $M - 0{,}5\sigma \leq x < M + 0{,}5\sigma$ | 3 (добре) |
| Вище середнього | $M + 0{,}5\sigma \leq x < M + 1{,}5\sigma$ | 4 (відмінно) |
| Високий | $x \geq M + 1{,}5\sigma$ | 5 (відмінно+) |

```python
# Межі нормувальної шкали
boundaries = {
    'Низький':              (0,           M - 1.5*sig),
    'Нижче середнього':     (M - 1.5*sig, M - 0.5*sig),
    'Середній':             (M - 0.5*sig, M + 0.5*sig),
    'Вище середнього':      (M + 0.5*sig, M + 1.5*sig),
    'Високий':              (M + 1.5*sig, 999)
}

print("=" * 70)
print("ТАБЛИЦЯ 3. Нормувальна шкала – динамометрія кисті (кгс)")
print(f"  M = {M:.2f} кгс,  σ = {sig:.2f} кгс")
print("=" * 70)
print(f"{'Рівень':<22} {'Нижня межа':>12} {'Верхня межа':>12} {'Оцінка':>8}")
print("-" * 70)

scores = [1, 2, 3, 4, 5]
labels_list = list(boundaries.keys())
for i, (level, (lo, hi)) in enumerate(boundaries.items()):
    lo_str = f"{lo:.1f}" if lo > 0 else "< -"
    hi_str = f"{hi:.1f}" if hi < 999 else "і вище"
    print(f"{level:<22} {lo_str:>12} {hi_str:>12} {scores[i]:>8}")

print("=" * 70)

# Присвоєння оцінок кожному спортсмену
def assign_score(x, M, sig):
    if   x < M - 1.5*sig:  return 1
    elif x < M - 0.5*sig:  return 2
    elif x < M + 0.5*sig:  return 3
    elif x < M + 1.5*sig:  return 4
    else:                   return 5

df['Оцінка'] = [assign_score(v, M, sig) for v in grip_best]
df['Рівень підготовленості'] = df['Оцінка'].map({
    1: 'Низький', 2: 'Нижче середнього',
    3: 'Середній', 4: 'Вище середнього', 5: 'Високий'
})

print("\nТАБЛИЦЯ 4. Індивідуальна оцінка силової підготовленості")
print("=" * 70)
print(df[['Спортсмен', 'Краща спроба (кгс)', 'Оцінка', 'Рівень підготовленості']].to_string(index=False))

print("\n" + "=" * 50)
print("Розподіл спортсменів за рівнями:")
print("=" * 50)
dist = df['Рівень підготовленості'].value_counts()
order = ['Низький', 'Нижче середнього', 'Середній', 'Вище середнього', 'Високий']
for level in order:
    count = dist.get(level, 0)
    pct   = count / len(df) * 100
    print(f"  {level:<22}: {count:2d} осіб  ({pct:.0f}%)")
```

## Завдання 3. Аналіз динамограми — імпульс, середня сила, градієнт

### Умова:
Записано **динамограму** (залежність сили F від часу t) при відштовхуванні спортсмена від опори. Визначити:
- Максимальну силу $F_{\max}$  
- Час досягнення максимуму $t_{F_{\max}}$  
- Імпульс сили $I = \int F(t)\,dt$  
- Середню силу $\bar{F} = I / T$  
- Градієнт сили $G_F = F_{\max} / t_{F_{\max}}$

```python
# ============================================================
# Модель динамограми відштовхування
# ============================================================

# Часова вісь (с)
T_total = 0.35           # загальний час фази відштовхування, с
dt      = 0.001
t       = np.arange(0, T_total + dt, dt)

# Маса тіла спортсмена
m_body  = 80             # кг
F_body  = m_body * 9.81  # сила тяжіння, Н

# Синтез реалістичної кривої F(t):
# фаза наростання (0 … t_peak), фаза зниження (t_peak … T_total)
t_peak = 0.12            # час до максимуму, с
F_max  = 2200            # максимальна сила, Н

F = np.where(
    t <= t_peak,
    F_body + (F_max - F_body) * np.sin(np.pi / 2 * t / t_peak)**1.6,
    F_body + (F_max - F_body) * np.exp(-4.5 * (t - t_peak) / (T_total - t_peak))
)
# Невеликий шум для реалістичності
np.random.seed(42)
F += np.random.normal(0, 15, size=len(t))
F = np.clip(F, 0, None)

# ============================================================
# РОЗРАХУНКИ
# ============================================================

F_max_val    = np.max(F)
t_Fmax       = t[np.argmax(F)]
F_mean       = np.mean(F)
impulse      = trapezoid(F, t)           # імпульс (Н·с)
F_avg_from_I = impulse / T_total         # середня сила через імпульс
gradient_F   = F_max_val / t_Fmax        # градієнт сили

# ============================================================
# ВИВЕДЕННЯ
# ============================================================
print("=" * 55)
print("РЕЗУЛЬТАТИ АНАЛІЗУ ДИНАМОГРАМИ")
print("=" * 55)
print(f"  Максимальна сила  F_max          = {F_max_val:.1f} Н")
print(f"  Час до максимуму  t(F_max)       = {t_Fmax:.3f} с")
print(f"  Загальний час     T              = {T_total:.3f} с")
print(f"  Імпульс сили      I = ∫F(t)dt   = {impulse:.2f} Н·с")
print(f"  Середня сила      F_avg = I/T    = {F_avg_from_I:.1f} Н")
print(f"  Градієнт сили     G_F = F/t(F)  = {gradient_F:.1f} Н/с")
print("=" * 55)
print(f"\n  Відносна сила     F_max / G_тіла = {F_max_val/F_body:.2f}")
print(f"  (F_тіла = {F_body:.1f} Н  при масі {m_body} кг)")
```

## Завдання 4. Визначення надійності тесту

Надійність тесту оцінюється **коефіцієнтом кореляції** між результатами першої та другої спроб:

$$r_{tt} = \frac{\sum(x_{1i} - \bar{x}_1)(x_{2i} - \bar{x}_2)}{\sqrt{\sum(x_{1i} - \bar{x}_1)^2 \cdot \sum(x_{2i} - \bar{x}_2)^2}}$$

**Критерії оцінки надійності:**

| $r_{tt}$ | Рівень надійності |
|----------|------------------|
| ≥ 0,90 | Відмінна |
| 0,80 – 0,89 | Хороша |
| 0,70 – 0,79 | Задовільна |
| < 0,70 | Незадовільна |

```python
# Коефіцієнт кореляції Пірсона
r_tt, p_value = stats.pearsonr(grip_1, grip_2)

# Значущість через t-критерій
n_s   = len(grip_1)
t_obs = r_tt * np.sqrt(n_s - 2) / np.sqrt(1 - r_tt**2)
t_cr  = stats.t.ppf(0.975, df=n_s - 2)   # критичне значення при α=0.05

# Стандартна похибка вимірювання
SEM_meas = np.std(grip_best, ddof=1) * np.sqrt(1 - r_tt)

print("=" * 60)
print("ОЦІНКА НАДІЙНОСТІ ТЕСТУ ДИНАМОМЕТРІЇ КИСТІ")
print("=" * 60)
print(f"  Кількість спортсменів  n    = {n_s}")
print(f"  Коефіцієнт кореляції   r_tt = {r_tt:.4f}")
print(f"  p-значення             p    = {p_value:.6f}")
print(f"  t-спостережуване       t    = {t_obs:.3f}")
print(f"  t-критичне (α=0.05)    t_cr = {t_cr:.3f}")
print(f"  Похибка вимірювання    SEM  = {SEM_meas:.2f} кгс")
print("-" * 60)

if   r_tt >= 0.90: rel_level = "ВІДМІННА"
elif r_tt >= 0.80: rel_level = "ХОРОША"
elif r_tt >= 0.70: rel_level = "ЗАДОВІЛЬНА"
else:              rel_level = "НЕЗАДОВІЛЬНА"

print(f"  Рівень надійності:  {rel_level}  (r_tt = {r_tt:.4f})")

if t_obs > t_cr:
    print(f"  Зв'язок СТАТИСТИЧНО ЗНАЧУЩИЙ (t={t_obs:.2f} > t_cr={t_cr:.2f})")
else:
    print(f"  Зв'язок НЕ значущий (t={t_obs:.2f} < t_cr={t_cr:.2f})")

print("=" * 60)
```

## Теоретичні питання для захисту

**Питання 1.** Що таке силові якості та яке їх значення в спорті?

> **Відповідь:** Силові якості — це здатність долати зовнішній опір або протидіяти йому за рахунок м'язових зусиль. Від рівня їх розвитку залежать досягнення практично у всіх видах спорту.

**Питання 2.** Назвіть три групи показників при контролі за силовими якостями.

> **Відповідь:** 1) **Основні** — максимальна та середня сила; 2) **Інтегральні** — імпульс сили; 3) **Диференціальні** — градієнт сили.

**Питання 3.** Що таке імпульс сили та як він розраховується?

> **Відповідь:** Імпульс сили — це інтеграл від сили по часу: $I = \int F(t)\,dt$ (або $I = F \cdot t$ при постійній силі). Він визначає кінцевий ефект дії сили на тіло (зміну кількості руху).

**Питання 4.** Що таке градієнт сили?

> **Відповідь:** Градієнт сили — диференціальний показник, що характеризує швидкість наростання сили: $G_F = dF/dt$ або практично $G_F = F_{\max}/t_{F_{\max}}$.

**Питання 5.** Яка надійність механічних динамометрів порівняно з тензометричними?

> **Відповідь:** Механічні динамометри мають надійність $r_{tt} = 0{,}60–0{,}80$, тензометричні пристрої — $r_{tt} = 0{,}85–0{,}95$, що значно вище.

**Питання 6.** Як побудувати нормувальну шкалу?

> **Відповідь:** На основі середнього $(M)$ та стандартного відхилення $(\sigma)$ вибірки будується 5-рівнева шкала з межами $M \pm 0{,}5\sigma$ (рівні 2–4) та $M \pm 1{,}5\sigma$ (рівні 1 і 5).
